### Welcome to the Apply Challenge

```python
link = 'https://raw.githubusercontent.com/chriszapp/datasets/main/books.csv'
books = pd.read_csv(link, nrows = 13)
books.head()
```

```python
# Firstly, have a look at the dataset
```

```python
# Mission 1: use apply to create a new column called 'reviews_opinion': if the ratings is between 4.5 and 5, the reviews_opinion is 'very good';
# if the rating is between 4 and 4.5, the reviews_opinion is 'good'; if the rating is between 3 and 4, the reviews_opinions will be 'neutral';
# if the rating is up to 3 the reviews_opinion is going to be 'bad'

# Add your code here

def opinion(rating):
    if rating >= 4.5:
        return "very good"
    elif rating >= 4:
        return "good"
    elif rating >= 3:
        return "neutral"
    else:
        return "bad"

books["reviews_opinion"] = books["average_rating"].apply(opinion)
books.head()
```

```python
# Mission 2: using apply with a function, create a column with the number of authors per book
# (note that in the column 'authors', each name of the author is divided with the character '/')

# Add your code here

def count_authors(authors):
    return len(authors.split("/"))       # "A/B" donne ['A', 'B'], et len compte 2

books["nb_authors"] = books["authors"].apply(count_authors)
books.head()
```

```python
# Mission 3: rewrite the same task as Mission 2 using the lamdba function

# Add your code here

books["nb_authors"] = books["authors"].apply(lambda authors: len(authors.split("/")))
books.head()
```

```python
# Mission 4: create a column 'comment_ration'.
# Use the columns 'ratings_count' and 'text_reviews_count' to calculate the percentage of how many text review per reviews are per book:
# percentage = text / reviews * 100
# Use the mathematical operations between columns

# Add your code here
books["comment_ration"] = books["text_reviews_count"] / books["ratings_count"] * 100
books.head()
```

```python
# Mission 5: obtain the same result of Mission 4 using apply with a function
# Try both methods to apply to the whole dataset and apply to selected columns

# Add your code here

# la fonction "ratio" attend de recevoir une ligne entière ("row").
def ratio(row):
    return row["text_reviews_count"] / row["ratings_count"] * 100       # Elle prend les deux chiffres de la ligne et calcule le pourcentage.


# Affiche la ligne complète, de toutes les colonnes.
books["comment_ration"] = books.apply(ratio, axis=1) 

# Affiche uniquement les deux colonnes choisies.
books["comment_ration"] = books[["ratings_count", "text_reviews_count"]].apply(ratio, axis=1)
```

```python
# Mission 6: obtain the same result of Mission 5 using apply lambda
# Try both methods to apply to the whole dataset and apply to selected columns

# Add your code here

# Calcule le pourcentage d'avis écrits par rapport au nombre total de notes pour chaque livre, puis enregisre le résultat dans une nouvelle colonne "comment_ration".
books["comment_ration"] = books.apply(lambda row: row["text_reviews_count"] / row["ratings_count"] * 100, axis=1)


# calcule le pourcentage d'avis écrits par rapport au nombre total de notes, tout en ne séléctionnant que les deux colonnes nécessaires.
books["comment_ration"] = books[["ratings_count", "text_reviews_count"]].apply(lambda row: row["text_reviews_count"] / row["ratings_count"] * 100, axis=1)
```

```python
# Mission 7: extract the year from the publication_date column. How many years of publications are recorded in this dataset? Which are them?
# Add your code here

books["year"] = pd.to_datetime(books["publication_date"], format="%m/%d/%Y").dt.year # Convertit la colonne en type date et crée une nouvelle colonne avec uniquement l'année

print(books["year"].nunique())
print(sorted(books["year"].unique()))
```
