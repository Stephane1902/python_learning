# Your mission is to determine which date it was / it will be

```python
# what month will it be in 7 weeks ?
from datetime import datetime
from datetime import timedelta
import pandas as pd

now = datetime.now()
print((now + timedelta(weeks=7)).month)
```

```python
# what day of the week will it be in 3 days ?

now = datetime.now()
print((now + timedelta(days=3)).weekday())
```

```python
# what day of the week will it be in 3 days ?

now = datetime.now()
print((now + timedelta(days=3)).strftime("%A"))
```

```python
# what hour will it be in 100 hours ?

now = datetime.now()
print((now + timedelta(hours=1000)).strftime("%H"))
```

```python
# what hour will it be in 100 hours ?

now = datetime.now()
print((now + timedelta(hours=1000)).hour)
```

```python
# what date was 30 days ago ?

now = datetime.now()
print((now - timedelta(days=30)).strftime("%d"))
```

```python
# what date was 30 days ago ?

now = datetime.now()
print((now - timedelta(days=30)))
```

```python
# Convert the "DATE" column into a datetime format, then display the same chart. Are ticks on X-axis better printed ?
from datetime import datetime
from datetime import timedelta
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Récupère les données du fichier .CSV
df = pd.read_csv("https://raw.githubusercontent.com/murpi/wilddata/master/quests/weather2019.csv")

# Affiche la date en format date, et non plus en foramt string
df["DATE"] = pd.to_datetime(df["DATE"])

# Configuration graphique du tableau de données
fig, ax = plt.subplots(figsize=(20, 10))

# Dessin des points sur le graphique
ax.scatter(x=df['DATE'],
           y=df['MAX_TEMPERATURE_C'])

# Affiche le résultats
fig.show()
```
