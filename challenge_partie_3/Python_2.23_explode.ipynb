- Which actor is mentioned the most often?


```python
import pandas as pd
import ast

link = 'https://raw.githubusercontent.com/justmarkham/pandas-videos/master/data/imdb_1000.csv'
df = pd.read_csv(link)

df["actors_list"] = df["actors_list"].apply(ast.literal_eval)       # Transforme le texte dela colonne en liste.

acteurs = df["actors_list"].explode().value_counts()                # Explose la liste en plusieurs lignes.

print(acteurs.head())
```
