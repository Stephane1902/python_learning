# PIVOT TABLE

## It's like in Excel, but with Pandas

There are 2 syntaxes, choose your preferred :
- As a method : YOUR_DATAFRAME.pivot_table()
- As a function : pd.crosstab()

```python
# Example
df_weather.pivot_table(values = 'MAX_TEMPERATURE_C', index = 'MONTH', aggfunc = 'mean')
```

## Create a Pivot table with :
- Months on rows
- the maximum of Max Temperature on values
- Opinion on columns

```python
# Hop !
df_weather.pivot_table(index = 'MONTH', columns = 'OPINION', values = 'MAX_TEMPERATURE_C', aggfunc = 'max')
```

## Pivot table with :
- Days on rows
- Precipitation on values
- Months on columns


```python
# Hop !
df_weather.pivot_table(index="DAY", values="PRECIP_TOTAL_DAY_MM", columns="MONTH", aggfunc="sum")
```

## Pivot table with :
- Months on rows
- the sum of Precipitations on values

**Then display it as a column chart.** (optionnal)

```python
# Hop !
rain = df_weather.pivot_table(index="MONTH", values="PRECIP_TOTAL_DAY_MM", aggfunc="sum")
rain.plot(kind="bar")      # kind & "bar" sont des mots clés basé sur Maplotib.
```

## Pivot table with :
- Months on rows
- the average of Max Temperature on a first column of values
- the average of Sunhours on a second column of values

```python
# Hop !
sun = df_weather.pivot_table(index = "MONTH", values = ["MAX_TEMPERATURE_C", "SUNHOUR"], aggfunc = "mean")
```

# Chart
Using your favorite library (MatPlotLib, Seaborn, Plotly or what you want), draw a chart from the previous pivot table with :
- Temperature in a blue column chart
- Sunhours in a red line plot

Tip : Once your pivot table created, you can help yourself with the `.reset_index()`  method.

```python
# Your code here
import matplotlib.pyplot as plt     # Import la librairie Matplotlib pour la visualisation des données.

pivot = sun.reset_index()           # Transforme l'index du tableau en une colonne normale, permettant de la placer sur l'axe horizontal.

fig, ax = plt.subplots()            # Prépare une feuille blanche ("fig") et un axe ("ax") pour le graphique.

ax.bar(pivot["MONTH"], pivot["MAX_TEMPERATURE_C"], color = "blue")          # Dessine un graphique en barres bleues : les mois en bas et les températures maximales à gauche

ax2 = ax.twinx()  # Crée un deuxième axe vertical partageant le même axe horizontal.

ax2.plot(pivot["MONTH"], pivot["SUNHOUR"], color = "red", marker = "o")    # Dessine une courbe rouge sur le deuxième axe : les mois en bas et les heures d'ensoleillement à droite

plt.show()  # Affiche le graphique final combinant les deux visualisations.
```
