# 1. Calcul d'un prix

Ecrivez un programme faisant saisir à l’utilisateur un prix unitaire HT, un taux de TVA et un nombre d’articles, et qui calcule le montant TTC de l’achat.<br>
Affichez le montant TTC.

```python
prix = float(input("Renseigne le prix unitaire HT : "))
tva = float(input("Renseigne le taux de TVA en % : "))
nombre_articles = int(input("Renseigne le nombre d'articles : "))

montant_ttc = prix * (1 + tva / 100) * nombre_articles

print("Le montant TTC est de :", montant_ttc, "€")
```

# 2. Conversion de durées (1)

Ecrivez un programme demandant une durée en secondes à l’utilisateur, et qui la convertit en heures, minutes, secondes.<br>
Affichez le résultat.<br>
Par exemple 12334 deviendra 3 heures, 25 minutes et 34 secondes.

```python
duree_en_secondes = int(input("renseigne la durée en secondes : "))

heures = duree_en_secondes // 3600
minutes = (duree_en_secondes - (heures * 3600)) // 60
secondes = duree_en_secondes % 60

print(duree_en_secondes, "deviendra", heures, "heure(s),", minutes, "minutes,", secondes, "secondes")
```

# 3. Conversion de durées (2)

Ecrivez un programme demandant une durée en heures, minutes, secondes à l’utilisateur et qui la convertit en secondes.<br>
Affichez le résultat.<br>
Par exemple 3 heures, 25 minutes et 34 secondes deviendra 12334 secondes.

```python
heures = int(input("renseigne le nombre d'heures : "))
minutes = int(input("renseigne le nombre de minutes : "))
secondes = int(input("renseigne le nombre de secondes : "))

heures = heures * 3600
minutes = minutes * 60
total_secondes = heures + minutes + secondes 

print("Le nombre total de secondes est : ", total_secondes)
```
