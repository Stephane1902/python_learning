# Mission 1

For each cell, execute the code and answer the question before moving on to the next.

```python
names = ["Fredy", "Nicolas", "Alexandre"]

for i in enumerate(names):
  print(i)
```

1. What type of object is returned as output?  
Hint: You can use `type(i)`.

Answer:


```python
# Un tuple (<class 'tuple'>) de 2 éléments : l'index  et l'élément. On le vérifie avec type(i).
```

```python
for i, val in enumerate(names):
  print(val)
```

2. What does `i` correspond to?  
What does `val` correspond to?  
Why is it so?

Answer:

```python
# i représente l'index (0,1,2)
# val représente les éléments ('Fredy','Nicolas','Alexandre')
# enumerate() renvoie des paires (index, élément), soit la boucle for index, élément in enumerate(): la première valeur va dans i, le deuxième dans  val
```

```python
for val, i in enumerate(names):
  print(val)
```

3. This time, `i` returns first names and `val` returns numbers. Why is it so?

Answer:

```python
# Le nom de la variable n'a pas d'importance, c'est uniquement la position qui détermine le type. val et i ayant inversés leurs positions, ils inversent aussi leurs types.
```

4. What does the built in function `enumerate` do?
In what cases can it be useful?

Answer:

```python
# La fonction enumerate() permet de renvoyer les éléments d'un itérable (listes, tuples, etc.) avec son index. Elle est utile lorsque l'on a besoin de la position et de la valeur dans une boucle.
```

```python
test_list = ['Hi', 'my', 'name', 'is', "hi", "my","name", "is", "chickychicky"]

test_list[8] = "Marshall"

test_list
```

1. What does the 8 in square brackets mean?  
What action has been performed?

Answer:

```python
# Le [8] réfère à la position de la valeur associée à l'index 8 de la liste. 
# test_list[8] = "Marshall" a remplacé "chickychicky" par "Marshall".
```

```python
for i in range(4):
  print(test_list[i])
```

2. What does `range(4)` mean?  
Why are only the first 4 elements of `test_list` displayed?

Answer:

```python
# range(4) génère les valeurs associées aux index 0,1,2,3 de i.
# i prend ces valeurs et sert d'index : seuls test_list[0] à test_list[3] sont affichés, soit les 4 premiers éléments.
```

```python
for i in range(2, 8, 2):
  print(test_list[i])
```

3. What does `range(2,8,2)` mean ?  
Why are only the elements 'name', 'hi' and 'name' displayed?

Answer:

```python
# range(start,end,pas) soit ici range indique un début à l'index 2, une fin à l'index 8 exclue, avec un pas de 2.
# Les index séléctionnés sont donc 2, 4, 6.
# Ces valeurs servent d'index : test_list[2] = 'name', test_list[4] = 'hi', test_list[6] = 'name'.
```
