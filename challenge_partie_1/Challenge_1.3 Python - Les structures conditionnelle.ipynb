# 1. Calcul du montant d'une remise

Un commerçant accorde une remise de 5 % pour tout achat d’un montant compris entre 100 et 500 € et 8 % au-delà.<br>
Ecrivez un programme de calcul du montant de la remise sur un achat donné.<br>
Affichez le montant de la remise.

```python
prix = float(input("renseigne le montant payé : "))

if prix > 500 : 
    print("la remise de 8 pourcent est accordée")
    remise = prix * 0.08
elif prix >= 100 :
    print("la remise de 5 pourcent est accordée")
    remise = prix * 0.05
else : 
    print("la remise n'est pas accordée")
    remise = 0

print("la remise est de : ", remise)
```

# 2. Savoir si trois entiers sont triés

Ecrivez un programme faisant saisir trois entiers x, y, z à l’utilisateur, et lui indiquer si ces nombres sont dans l’ordre croissant (x <= y <= z).

```python
x = int(input("renseigne un nombre entier : "))
y = int(input("renseigne un autre nombre entier : "))
z = int(input("renseigne un dernier nombre entier : "))

if x <= y and y <= z :
    print("Ces nombres sont dans l'ordre croissant")
else :
    print("Ces nombres ne sont pas dans l'ordre croissant") 
```

# 3. Tri de trois réels

Ecrivez un programme faisant saisir trois nombres réels x, y, z à l’utilisateur et qui les trie par ordre croissant (à la fin du déroulement du programme x ≤ y ≤ z).<br>
Affichez x, y et z.

```python
x = float(input("renseigne un nombre réel : "))
y = float(input("renseigne un autre nombre réel : "))
z = float(input("renseigne un dernier nombre réel : "))

if x > y :
    t = x   # t = variable temporaire 
    x = y
    y = t

if y > z :
    t = y
    y = z
    z = t

if x > y :
    t = x
    x = y
    y = t

print("Les nombres dans l'ordre croissant sont : ", x, y, z)
```

# 4. Signe d'un produit

Ecrivez un programme qui affiche le signe du produit de deux nombres réels sans calculer la valeur de ce produit. Par signe, on entend positif, négatif ou nul.

```python
a = float(input("renseigne un nombre : "))
b = float(input("renseigne un autre nombre : "))

if a > 0 and b > 0 :
    print("le produit de ces deux nombres est positif")
elif a < 0 and b < 0 :
    print("le produit de ces deux nombres est positif")
elif a < 0 and b > 0 or a > 0 and b < 0 :
    print("le produit de ces deux nombres est négatif")
else :
    print("le produit de ces deux nombres est nul")
```
