# Mission 2

Create a "fruit_vendors" dataframe that reflects the below example **using a dictionary** data structure

![](https://i.imgur.com/CHPn7ZF.png)

[This resource could help you.](https://pandas.pydata.org/pandas-docs/stable/reference/api/pandas.DataFrame.from_dict.html)

```python
import pandas as pd

fruit_vendors = pd.DataFrame([[35, 21], [41, 34]],
    columns=["Apples", "Bananas"],
    index=["2017 Sales", "2018 Sales"])
print(fruit_vendors)
```

```python
import pandas as pd

fruit_vendors = pd.DataFrame({"": ["2017 Sales", "2018 Sales"],
                              "Apples" : [35, 21],
                              "Bananas" : [41, 34]})
print(fruit_vendors)
```

```python
import pandas as pd

fruit_vendors = pd.DataFrame({"Apples": [35, 41], "Bananas": [21, 34]}, index=["2017 Sales", "2018 Sales"])

print(fruit_vendors)
```

# Mission 3

Use a loop to display the values from each column in the dataframe from mission 2.

[This resource could help you](https://www.w3schools.com/python/pandas/ref_df_items.asp)

```python
import pandas as pd

for columns, elements in fruit_vendors.items():
    print(columns)
    print(elements)
```
